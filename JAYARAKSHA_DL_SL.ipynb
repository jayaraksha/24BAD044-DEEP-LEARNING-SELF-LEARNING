```python
!pip install -q gensim transformers sentencepiece

import pandas as pd
import numpy as np
import re
import time
import matplotlib.pyplot as plt
import tensorflow as tf
import torch
from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.decomposition import PCA, TruncatedSVD
from sklearn.metrics.pairwise import cosine_similarity
from gensim.models import Word2Vec, FastText
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GlobalAveragePooling1D, Dense
from transformers import AutoTokenizer, AutoModel

# Load IMDB dataset automatically

num_words = 10000
(train_data, train_labels), (test_data, test_labels) = tf.keras.datasets.imdb.load_data(
    num_words=num_words
)
word_index = tf.keras.datasets.imdb.get_word_index()

reverse_word_index = {
    value + 3: key for key, value in word_index.items()
}
reverse_word_index[0] = "<PAD>"
reverse_word_index[1] = "<START>"
reverse_word_index[2] = "<UNK>"
reverse_word_index[3] = "<UNUSED>"
def decode_review(encoded_review):
    return " ".join(
        reverse_word_index.get(i, "?")
        for i in encoded_review
    )
train_texts = [
    decode_review(review)
    for review in train_data
]
test_texts = [
    decode_review(review)
    for review in test_data
]
texts = train_texts
print("Dataset loaded successfully")
print("Training samples:", len(train_texts))
print("Testing samples:", len(test_texts))
print("Total samples:", len(train_texts) + len(test_texts))
print("\nFirst review:")
print(train_texts[0][:1000])

# Text cleaning

def clean_text(text):
    text = text.lower()
    text = re.sub(
        r"<br\s*/?>",
        " ",
        text
    )
    text = re.sub(
        r"[^a-z\s]",
        " ",
        text
    )
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text
texts = [
    clean_text(text)
    for text in texts
]
print("\nCleaned review:")
print(texts[0][:1000])

# Tokenization

tokenized_texts = [
    re.findall(
        r"\b[a-z]+\b",
        text
    )
    for text in texts
]
print("\nTokenized review:")
print(tokenized_texts[0][:30])

# One-Hot Encoding

sample_words = [
    "movie",
    "film",
    "good",
    "bad",
    "great"
]
try:
    encoder = OneHotEncoder(
        sparse_output=False
    )
except:
    encoder = OneHotEncoder(
        sparse=False
    )
one_hot = encoder.fit_transform(
    np.array(sample_words).reshape(-1, 1)
)
one_hot_df = pd.DataFrame(
    one_hot,
    index=sample_words,
    columns=encoder.get_feature_names_out(["word"])
)
print("\nOne-Hot Encoding:")
display(one_hot_df)

# Bag of Words

bow_vectorizer = CountVectorizer(
    max_features=5000
)
bow_matrix = bow_vectorizer.fit_transform(
    texts
)
print("\nBag of Words shape:")
print(bow_matrix.shape)
print("\nBoW vocabulary:")
print(
    bow_vectorizer.get_feature_names_out()[:30]
)

# N-Grams

ngram_vectorizer = CountVectorizer(
    max_features=5000,
    ngram_range=(1, 2)
)
ngram_matrix = ngram_vectorizer.fit_transform(
    texts
)
print("\nN-Gram matrix shape:")
print(ngram_matrix.shape)

print("\nSample N-Grams:")
print(
    ngram_vectorizer.get_feature_names_out()[:30]
)

# TF-IDF

tfidf_vectorizer = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    stop_words="english"
)
tfidf_matrix = tfidf_vectorizer.fit_transform(
    texts
)
print("\nTF-IDF matrix shape:")
print(tfidf_matrix.shape)
tfidf_features = (
    tfidf_vectorizer
    .get_feature_names_out()
)
print("\nTF-IDF vocabulary:")
print(tfidf_features[:30])
tfidf_sample = pd.DataFrame(
    tfidf_matrix[:5].toarray(),
    columns=tfidf_features
)
display(
    tfidf_sample.iloc[:, :20]
)

# LSA using SVD

svd = TruncatedSVD(
    n_components=50,
    random_state=42
)

lsa_matrix = svd.fit_transform(
    tfidf_matrix
)

print("\nLSA matrix shape:")
print(lsa_matrix.shape)

print(
    "Explained variance:",
    round(
        svd.explained_variance_ratio_.sum(),
        4
    )
)


# Word2Vec

start_time = time.time()

word2vec_model = Word2Vec(
    sentences=tokenized_texts,
    vector_size=100,
    window=5,
    min_count=5,
    workers=4,
    sg=1,
    epochs=10
)

word2vec_time = time.time() - start_time

print("\nWord2Vec completed")

print(
    "Vocabulary size:",
    len(word2vec_model.wv.key_to_index)
)

print(
    "Training time:",
    round(word2vec_time, 2),
    "seconds"
)


# Word2Vec vector

query_word = "movie"

if query_word in word2vec_model.wv:

    vector = word2vec_model.wv[
        query_word
    ]

    print("\nWord:", query_word)
    print(
        "Vector shape:",
        vector.shape
    )

    print(
        "First 20 values:",
        vector[:20]
    )


# Word2Vec semantic similarity

query_words = [
    "movie",
    "good",
    "film",
    "great"
]

print("\nFive most similar words")

for query_word in query_words:

    if query_word in word2vec_model.wv:

        print(
            "\nSimilar words for:",
            query_word
        )

        similar_words = (
            word2vec_model.wv
            .most_similar(
                query_word,
                topn=5
            )
        )

        for word, score in similar_words:

            print(
                word,
                ":",
                round(score, 4)
            )


# Cosine similarity

word1 = "movie"
word2 = "film"

if (
    word1 in word2vec_model.wv
    and
    word2 in word2vec_model.wv
):

    vector1 = (
        word2vec_model.wv[word1]
        .reshape(1, -1)
    )

    vector2 = (
        word2vec_model.wv[word2]
        .reshape(1, -1)
    )

    similarity = cosine_similarity(
        vector1,
        vector2
    )[0][0]

    print(
        "\nCosine similarity between",
        word1,
        "and",
        word2,
        ":",
        round(similarity, 4)
    )


# FastText

start_time = time.time()

fasttext_model = FastText(
    sentences=tokenized_texts,
    vector_size=100,
    window=5,
    min_count=5,
    workers=4,
    epochs=10
)

fasttext_time = time.time() - start_time

print("\nFastText completed")

print(
    "Vocabulary size:",
    len(fasttext_model.wv.key_to_index)
)

print(
    "Training time:",
    round(fasttext_time, 2),
    "seconds"
)


# FastText semantic similarity

print("\nFastText similar words")

for query_word in query_words:

    print(
        "\nSimilar words for:",
        query_word
    )

    similar_words = (
        fasttext_model.wv
        .most_similar(
            query_word,
            topn=5
        )
    )

    for word, score in similar_words:

        print(
            word,
            ":",
            round(score, 4)
        )


# FastText OOV demonstration

unseen_word = "moviezz"

print("\nFastText OOV demonstration")

print(
    "Is word in vocabulary?",
    unseen_word in fasttext_model.wv.key_to_index
)

try:

    unseen_vector = (
        fasttext_model.wv[unseen_word]
    )

    print(
        "Vector generated successfully"
    )

    print(
        "Vector shape:",
        unseen_vector.shape
    )

    print(
        "First 10 values:",
        unseen_vector[:10]
    )

except Exception as e:

    print(
        "Error:",
        e
    )


# Keras Tokenization

MAX_WORDS = 10000
MAX_LENGTH = 200

keras_tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token="<OOV>"
)

keras_tokenizer.fit_on_texts(
    texts
)

sequences = (
    keras_tokenizer
    .texts_to_sequences(texts)
)

padded_sequences = pad_sequences(
    sequences,
    maxlen=MAX_LENGTH,
    padding="post",
    truncating="post"
)

vocab_size = min(
    MAX_WORDS,
    len(
        keras_tokenizer.word_index
    ) + 1
)

print("\nKeras Embedding")

print(
    "Padded sequence shape:",
    padded_sequences.shape
)

print(
    "Vocabulary size:",
    vocab_size
)


# Keras Embedding Model

keras_model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=64
    ),
    GlobalAveragePooling1D(),
    Dense(
        64,
        activation="relu"
    ),
    Dense(
        1,
        activation="sigmoid"
    )
])

keras_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

keras_model.build(
    input_shape=(None, MAX_LENGTH)
)

keras_model.summary()


# Prepare labels

labels = np.array(
    train_labels
)


# Train Keras Embedding

history = keras_model.fit(
    padded_sequences,
    labels,
    epochs=3,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)


# Keras Embedding vectors

embedding_layer = (
    keras_model.layers[0]
)

embedding_weights = (
    embedding_layer
    .get_weights()[0]
)

print(
    "\nEmbedding matrix shape:",
    embedding_weights.shape
)

keras_word_index = (
    keras_tokenizer.word_index
)

selected_words = [
    "movie",
    "film",
    "good",
    "bad",
    "great",
    "love",
    "horrible"
]

for word in selected_words:

    if word in keras_word_index:

        index = keras_word_index[word]

        if index < vocab_size:

            print(
                "\nWord:",
                word
            )

            print(
                "Embedding:",
                embedding_weights[index][:10]
            )


# BERT

bert_tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

bert_model = AutoModel.from_pretrained(
    "bert-base-uncased"
)

bert_model.eval()

print("\nBERT model loaded")


# BERT sentence embedding

sentence = (
    "The movie was very good."
)

inputs = bert_tokenizer(
    sentence,
    return_tensors="pt"
)

with torch.no_grad():

    outputs = bert_model(
        **inputs
    )

last_hidden_state = (
    outputs.last_hidden_state
)

print(
    "\nSentence:",
    sentence
)
print(
    "BERT output shape:",
    last_hidden_state.shape
)
attention_mask = inputs[
    "attention_mask"
]
mask = attention_mask.unsqueeze(
    -1
)
sentence_embedding = (
    last_hidden_state * mask
).sum(dim=1) / mask.sum(dim=1)

print(
    "Sentence embedding shape:",
    sentence_embedding.shape
)
print(
    "First 10 values:",
    sentence_embedding[0][:10]
)
# BERT contextual representation
context_sentences = [
    "The bank approved the loan.",
    "We sat near the bank of the river."
]
def get_word_embedding(
    sentence,
    target_word
):
    inputs = bert_tokenizer(
        sentence,
        return_tensors="pt"
    )
    with torch.no_grad():

        outputs = bert_model(
            **inputs
        )
    tokens = (
        bert_tokenizer
        .convert_ids_to_tokens(
            inputs["input_ids"][0]
        )
    )
    target_indices = []
    for i, token in enumerate(tokens):
        if token == target_word:
            target_indices.append(i)
    if len(target_indices) == 0:
        return None, tokens
    embeddings = (
        outputs
        .last_hidden_state[0]
    )
    target_embedding = (
        embeddings[target_indices]
        .mean(dim=0)
    )
    return (
        target_embedding,
        tokens
    )
bank_embeddings = []
for sentence in context_sentences:
    embedding, tokens = (
        get_word_embedding(
            sentence,
            "bank"
        )
    )
    print(
        "\nSentence:",
        sentence
    )
    print(
        "Tokens:",
        tokens
    )
    if embedding is not None:
        print(
            "Bank embedding shape:",
            embedding.shape
        )
        bank_embeddings.append(
            embedding.numpy()
        )
if len(bank_embeddings) == 2:
    contextual_similarity = (
        cosine_similarity(
            [bank_embeddings[0]],
            [bank_embeddings[1]]
        )[0][0]
    )
    print(
        "\nCosine similarity between",
        "the two contextual bank representations:",
        round(
            contextual_similarity,
            4
        )
    )
# Word2Vec PCA visualization
available_words = (
    word2vec_model
    .wv
    .index_to_key[:50]
)
vectors = np.array([
    word2vec_model.wv[word]
    for word in available_words
])
pca = PCA(
    n_components=2,
    random_state=42
)
reduced_vectors = (
    pca.fit_transform(vectors)
)
plt.figure(
    figsize=(14, 10)
)
plt.scatter(
    reduced_vectors[:, 0],
    reduced_vectors[:, 1]
)
for i, word in enumerate(
    available_words
):
    plt.annotate(
        word,
        (
            reduced_vectors[i, 0],
            reduced_vectors[i, 1]
        )
    )
plt.title(
    "Word2Vec Embeddings Using PCA"
)
plt.xlabel(
    "Principal Component 1"
)

plt.ylabel(
    "Principal Component 2"
)

plt.grid()

plt.show()


# Keras Embedding PCA

keras_words = []

keras_vectors = []

for word in selected_words:

    if word in keras_word_index:

        index = keras_word_index[word]

        if index < vocab_size:

            keras_words.append(word)

            keras_vectors.append(
                embedding_weights[index]
            )

keras_vectors = np.array(
    keras_vectors
)

if len(keras_vectors) >= 2:

    pca = PCA(
        n_components=2,
        random_state=42
    )

    reduced = pca.fit_transform(
        keras_vectors
    )

    plt.figure(
        figsize=(10, 7)
    )

    plt.scatter(
        reduced[:, 0],
        reduced[:, 1]
    )

    for i, word in enumerate(
        keras_words
    ):

        plt.annotate(
            word,
            (
                reduced[i, 0],
                reduced[i, 1]
            )
        )

    plt.title(
        "Keras Embeddings Using PCA"
    )

    plt.xlabel(
        "Principal Component 1"
    )

    plt.ylabel(
        "Principal Component 2"
    )

    plt.grid()

    plt.show()


# TensorFlow Embedding Projector files

projector_words = (
    word2vec_model
    .wv
    .index_to_key[:200]
)

projector_vectors = np.array([
    word2vec_model.wv[word]
    for word in projector_words
])

np.savetxt(
    "word2vec_vectors.tsv",
    projector_vectors,
    delimiter="\t"
)

with open(
    "word2vec_metadata.tsv",
    "w"
) as f:

    for word in projector_words:

        f.write(
            word + "\n"
        )

print(
    "\nEmbedding Projector files created:"
)

print(
    "word2vec_vectors.tsv"
)

print(
    "word2vec_metadata.tsv"
)

print(
    "\nOpen:"
)

print(
    "https://projector.tensorflow.org/"
)


# Comparison table

comparison = pd.DataFrame({
    "Technique": [
        "One-Hot Encoding",
        "Bag of Words",
        "TF-IDF",
        "LSA",
        "Word2Vec",
        "FastText",
        "Keras Embedding",
        "BERT"
    ],

    "Representation": [
        "Sparse binary vector",
        "Sparse word-frequency vector",
        "Sparse weighted vector",
        "Dense reduced vector",
        "Dense word vector",
        "Dense subword vector",
        "Trainable dense vector",
        "Contextual dense vector"
    ],

    "Contextual": [
        "No",
        "No",
        "No",
        "No",
        "No",
        "No",
        "Depends on model",
        "Yes"
    ],

    "OOV Handling": [
        "Poor",
        "Poor",
        "Poor",
        "Poor",
        "Limited",
        "Good",
        "Limited",
        "Subword tokenization"
    ],

    "Semantic Understanding": [
        "Very Low",
        "Low",
        "Low",
        "Moderate",
        "High",
        "High",
        "Learned",
        "Very High"
    ],

    "Computational Requirement": [
        "Very Low",
        "Low",
        "Low",
        "Moderate",
        "Moderate",
        "Moderate",
        "Moderate",
        "High"
    ]

})
display(comparison)


print(
    "\nComparison Table"
)

display(
    comparison
)


# Word2Vec results table

similarity_results = []

for query_word in [
    "movie",
    "good",
    "film",
    "great"
]:

    if query_word in word2vec_model.wv:

        similar = (
            word2vec_model
            .wv
            .most_similar(
                query_word,
                topn=5
            )
        )

        for rank, (
            word,
            score
        ) in enumerate(
            similar,
            start=1
        ):

            similarity_results.append({

                "Query Word": query_word,

                "Rank": rank,

                "Similar Word": word,

                "Cosine Similarity": round(
                    score,
                    4
                )
            })

similarity_df = pd.DataFrame(
    similarity_results
)

print(
    "\nWord2Vec Similarity Results"
)

display(
    similarity_df
)


# FastText results table

fasttext_results = []

for query_word in [
    "movie",
    "good",
    "film",
    "great"
]:

    similar = (
        fasttext_model
        .wv
        .most_similar(
            query_word,
            topn=5
        )
    )

    for rank, (
        word,
        score
    ) in enumerate(
        similar,
        start=1
    ):

        fasttext_results.append({

            "Query Word": query_word,

            "Rank": rank,

            "Similar Word": word,

            "Cosine Similarity": round(
                score,
                4
            )
        })

fasttext_df = pd.DataFrame(
    fasttext_results
)

print(
    "\nFastText Similarity Results"
)

display(
    fasttext_df
)


# Training time comparison

training_time_df = pd.DataFrame({

    "Model": [
        "Word2Vec",
        "FastText"
    ],

    "Training Time (seconds)": [
        round(
            word2vec_time,
            2
        ),
        round(
            fasttext_time,
            2
        )
    ]
})

print(
    "\nTraining Time Comparison"
)

display(
    training_time_df
)


print(
    "\nAssignment implementation completed successfully!"
)
```
